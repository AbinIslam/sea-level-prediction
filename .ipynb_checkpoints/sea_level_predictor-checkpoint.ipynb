```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import linregress
```

```python
sea_level_data = pd.read_csv("epa-sea-level.csv")

#sea_level_data
```

```python
CSIRO_sea_level = sea_level_data[['Year', 'CSIRO Adjusted Sea Level']]
CSIRO_sea_level = CSIRO_sea_level.dropna()
#CSIRO_sea_level
```

```python
year_extended = np.arange(1880, 2050, 1)
```

```python
y = CSIRO_sea_level['CSIRO Adjusted Sea Level'] #y
x = CSIRO_sea_level['Year'] #x
```

```python
#Linear regression
slope, intercept, r, p, std_err = linregress(x, y)
```

```python
line = [slope *xi + intercept for xi in year_extended]
```

```python
def predict_sea_level(x):
    return slope * x + intercept
```

```python
predict_model = list(map(predict_sea_level, x))
```

```python
sea_level = predict_sea_level(2050)
sea_level #Shows the actual sea level
```

```python
plt.scatter(x = 'Year', y = 'CSIRO Adjusted Sea Level', s = 4, color = 'blue', label = 'Data Points', data = CSIRO_sea_level)
plt.plot(year_extended, line, color = 'red', label = f'Best fit line: y={slope:.2f}x+{intercept:.2f}')
plt.xlabel('Year')
plt.ylabel('Sea Level (inches)')
plt.title("Rising Sea Level")
plt.legend()
plt.show()
```
